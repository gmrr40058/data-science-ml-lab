<h2>Dimensions in Arrays</h2>

<h3>0-D Arrays</h3>

0-D arrays, or Scalars, are the elements in an array. Each value in an array is a 0-D array.

```python
#Create a 0-D array with value 42
zero_D_array=np.array(42)
```

```python
zero_D_array
```

<h3>1-D Arrays</h3>

An array that has 0-D arrays as its elements is called uni-dimensional or 1-D array.<br>
These are the most common and basic arrays.

```python
#Create a 1-D array containing the values 1,2,3,4,5:
one_D_array=np.array([1,2,3,4,5])
```

```python
one_D_array
```

<h3>2-D Arrays</h3>

An array that has 1-D arrays as its elements is called a 2-D array.<br>
These are often used to represent matrix or 2nd order tensors.<br><br>
<mark>*NumPy has a whole sub module dedicated towards matrix operations called numpy.mat*</mark>

```python
#Create a 2-D array containing two arrays with the values 1,2,3 and 4,5,6:
two_D_array=np.array([[1, 2, 3], [4, 5, 6]])
```

```python
two_D_array
```

<h3>3-D arrays</h3>

An array that has 2-D arrays (matrices) as its elements is called 3-D array.<br>
These are often used to represent a 3rd order tensor.

```python
#Create a 3-D array with two 2-D arrays, both containing two arrays with the values 1,2,3 and 4,5,6:
three_D_array=np.array([[[1, 2, 3], [4, 5, 6]], [[1, 2, 3], [4, 5, 6]]])
```

```python
three_D_array
```

<h3>Check Number of Dimensions?</h3>

NumPy Arrays provides the ndim attribute that returns an integer that tells us how many dimensions the array have.

```python
zero_D_array.ndim
```

```python
one_D_array.ndim
```

```python
two_D_array.ndim
```

```python
three_D_array.ndim
```

<h3>Higher Dimensional Arrays</h3>

An array can have any number of dimensions.<br>
When the array is created, you can define the number of dimensions by using the ndmin argument.

```python
#Create an array with 5 dimensions and verify that it has 5 dimensions:
five_D_array = np.array([1, 2, 3, 4], ndmin=5) # Follow the argument name carefully, it is ndmin not ndim
```

```python

five_D_array
```

```python

five_D_array.ndim
```
