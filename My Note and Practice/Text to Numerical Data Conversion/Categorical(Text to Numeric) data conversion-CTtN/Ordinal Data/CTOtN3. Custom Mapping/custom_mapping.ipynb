```python
import pandas as pd
```

```python
data = {'Size': ['Medium', 'Small', 'Large', 'Medium', 'Large']}
df = pd.DataFrame(data)
```

```python
df
```

```python
mapping = {
    'Small' : 10,
    'Medium' : 11,
    'Large' : 12
}
```

```python
df['Size_encoded'] = df['Size'].map(mapping)
```

```python
df
```

We can do it using replace also

```python
df['Size_encoded_re'] = df['Size'].replace(mapping)
```

```python
df
```

<div style='text-align: center; font-size: 40px'> Difference between map()and replace()</div>

```python
data = {'Size': ['Medium', 'Small', 'Large', 'Medium', 'Large', '', 'Extra Large']}
df = pd.DataFrame(data)
```

```python
df
```

```python
mapping = {
    'Small' : 10,
    'Medium' : 11,
    'Large' : 12
}
```

```python
df['Size_encoded_map'] = df['Size'].map(mapping)
```

```python
df
```

```python
df['Size_encoded_rep'] = df['Size'].replace(mapping)
```

```python
df
```

##### For map()
**Notice:**
* Extra Large” is NOT in the mapping (6).
* There is a black cell in the column (5).

**.map() converts unknowns and blank to NaN**
##### For replace()
**.replace() does NOT convert unknown values and blank to NaN.
It simply leaves them unchanged.**
**
##### 🔥 Super simple summary
* .map() → unknown values → NaN
* .replace() → unknown values → keep original

<div style='text-align: center; font-size: 40px;'>Custom mapping on multiple column</div>

```python

# Sample DataFrame
df = pd.DataFrame({
    "Size": ["Small", "Medium", "Large", "Medium"],
    "Education": ["High School", "Bachelor", "Master", "PhD"],
    "Satisfaction": ["Low", "Medium", "High", "Very High"]
})
```

```python
df
```

```python

# Custom mappings for each ordinal column
size_map = {
    "Small": 0,
    "Medium": 1,
    "Large": 2
}

education_map = {
    "High School": 0,
    "Bachelor": 1,
    "Master": 2,
    "PhD": 3
}

satisfaction_map = {
    "Low": 0,
    "Medium": 1,
    "High": 2,
    "Very High": 3
}
```

```python
# Apply custom mapping (column by column)
df["Size"] = df["Size"].map(size_map)
df["Education"] = df["Education"].map(education_map)
df["Satisfaction"] = df["Satisfaction"].map(satisfaction_map)
```

```python
df
```
