```python
import pandas as pd
```

<div style='text-align: center; font-size: 40px'>Basic target encoding with pandas (no smoothing)</div>

```python
df = pd.DataFrame({
    "Color": ["Red","Red","Blue","Blue","Green","Yellow"],
    "target": [1,0,1,1,0,1]
})
```

```python
te = df.groupby('Color')['target'].mean()
te
```

```python
df["Color_te_basic"] = df["Color"].map(te)
df
```
