```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = x**2

plt.plot(x, y)          # line
plt.title("Single plot using plt.plot")
plt.xlabel("x")
plt.ylabel("x^2")

plt.show()
```

```python
x = np.linspace(0, 10, 100)

# First figure
plt.figure(figsize=(5, 3))
plt.plot(x, np.sin(x))
plt.title("Figure 1: sin(x)")

# Second figure
plt.figure(figsize=(5, 3))
plt.plot(x, np.cos(x), color="orange")
plt.title("Figure 2: cos(x)")

plt.show()
```

```python
x = np.linspace(0, 10, 100)

# First figure
plt.figure(figsize=(5, 3))
plt.plot(x, np.sin(x))
plt.plot(x, np.cos(x), color="orange")
plt.title("sin & cos on one pltes")
plt.xlabel("x")
plt.ylabel("value")
plt.legend()
plt.show()
```

```python
x = np.linspace(0, 10, 100)

fig, ax = plt.subplots(figsize=(5, 3))

ax.plot(x, np.sin(x), label="sin(x)")
ax.plot(x, np.cos(x), label="cos(x)")
ax.set_title("sin & cos on one Axes")
ax.set_xlabel("x")
ax.set_ylabel("value")
ax.legend()

plt.show()
```

```python
epochs = np.arange(1, 11)
train_loss = np.linspace(1.0, 0.2, 10)
val_loss = np.linspace(1.2, 0.3, 10)

fig, axs = plt.subplots(1, 2, figsize=(10, 4))

# Left plot: training loss
axs[0].plot(epochs, train_loss, marker="o")
axs[0].set_title("Training Loss")
axs[0].set_xlabel("Epoch")
axs[0].set_ylabel("Loss")

# Right plot: validation loss
axs[1].plot(epochs, val_loss, marker="o", color="orange")
axs[1].set_title("Validation Loss")
axs[1].set_xlabel("Epoch")
axs[1].set_ylabel("Loss")

fig.suptitle("Model Loss Curves")

plt.tight_layout()
plt.show()
```

```python
x = np.linspace(0, 10, 100)

# 1 row, 2 columns, first subplot
plt.subplot(2, 2, 1)
plt.plot(x, np.sin(x))
plt.xlabel("x")
plt.ylabel("sin x")
plt.title("sin(x)")

# 1 row, 2 columns, second subplot
plt.subplot(2, 2, 2)
plt.plot(x, np.sinh(x))
plt.title("sinh(x)")
plt.xlabel("x")
plt.ylabel("sinh x")

plt.subplot(2, 2, 3)
plt.plot(x, np.cos(x))
plt.title("cos(x)")
plt.xlabel("x")
plt.ylabel("cos x")

plt.subplot(2, 2, 4)
plt.plot(x, np.cosh(x))
plt.title("cosh(x)")
plt.xlabel("x")
plt.ylabel("cosh x")

plt.tight_layout()
plt.show()
```

```python
x = np.linspace(0, 10, 100)

fig, axs = plt.subplots(2, 2, figsize=(8, 6))

axs[0, 0].plot(x, np.sin(x))
axs[0, 0].set_title("sin(x)")

axs[0, 1].plot(x, np.cos(x), color="orange")
axs[0, 1].set_title("cos(x)")

axs[1, 0].plot(x, np.tan(x))
axs[1, 0].set_title("tan(x)")

axs[1, 1].plot(x, np.sinh(x), color="green")
axs[1, 1].set_title("sinh(x)")

for ax in axs.flat:
    ax.set_xlabel("x")
    ax.set_ylabel("value")

fig.suptitle("2x2 grid using subplots", fontsize=14)
plt.tight_layout()
plt.show()
```

```python
x = np.linspace(0, 10, 100)

# 1 row, 2 columns, first subplot
plt.subplot(1, 2, 1)
plt.plot(x, np.sin(x))
plt.plot(x, np.cos(x))
plt.xlabel("x")
plt.ylabel("value")
plt.title("sin(x) & cos(x)")

# 1 row, 2 columns, second subplot
plt.subplot(2, 2, 2)
plt.plot(x, np.cos(x))
plt.title("cos(x)")
plt.xlabel("x")
plt.ylabel("cos x")

plt.subplot(2, 2, 4)
plt.plot(x, np.tan(x))
plt.title("tan(x)")
plt.xlabel("x")
plt.ylabel("tan x")

plt.tight_layout()
plt.show()
```

```python
x = np.linspace(0, 10, 100)

fig, axs = plt.subplot_mosaic(
    [["left", "right_top"],
     ["left", "right_bottom"]],
    figsize=(8, 5)
)

axs["left"].plot(x, np.sin(x), label="sin(x)")
axs["left"].plot(x, np.cos(x), label="cos(x)")
axs["left"].set_title("Main: sin & cos")
axs["left"].legend()

axs["right_top"].plot(x, np.sin(x)**2, color="purple")
axs["right_top"].set_title("sin^2(x)")

axs["right_bottom"].plot(x, np.cos(x)**2, color="red")
axs["right_bottom"].set_title("cos^2(x)")

fig.suptitle("Custom layout with subplot_mosaic")
plt.tight_layout()
plt.show()
```
