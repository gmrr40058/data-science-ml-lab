<h2 style="text-align: center;">01:20:25 - [What is an IDE]</h2>

An **IDE** stands for **Integrated Development Environment**.

It is a software application that provides comprehensive tools for computer programmers to develop software. Typically, an IDE consists of at least three core components:

* **Source Code Editor:** A text editor designed specifically for writing and editing code, often featuring syntax highlighting and auto-completion.  
* **Build Automation Tools:** Utilities to automate the process of compiling source code into executable binaries or running scripts.  
* **Debugger:** A program used to test and debug other programs, allowing developers to inspect code, set breakpoints, and track down errors.

Some IDE’s are as follows:

1. **Jupyter Notebook:** A popular web-based interactive computing environment for creating and sharing documents.  
2. **PyCharm:** A dedicated Python IDE providing intelligent code completion and robust debugging features.  
3. **Visual Studio Code (VS Code):** A versatile, lightweight code editor with an extensive ecosystem of extensions for debugging and development.  
4. **Spyder:** An open-source scientific environment designed for data analysis and numerical computing.  
5. **Canopy:** A Python development platform built to streamline scientific and analytical workflows.  
6. **Vim:** A highly configurable text editor designed for efficient text manipulation and script editing.  
7. **Atom:** A customizable open-source text editor supporting various programming tools and plugins.

```python
"""
PURPOSE:
By default, VS Code's "Run All" feature halts execution at the first cell that raises an error.
Because this notebook contains intentional errors for learning and documentation purposes,
standard execution stops prematurely, requiring subsequent cells to be run manually.

HOW THIS HELPS:
This code defines and registers a custom IPython cell magic (`%%allow_error`).
When placed at the very top of an intentional error cell, it:
1. Executes the cell via IPython's internal runner (`run_cell`).
2. Prints the full, native traceback directly in the output area.
3. Swallows the exception at the notebook runner level so VS Code marks the cell as complete
   and allows "Run All" to continue running subsequent cells uninterrupted.

USAGE:
Add `%%allow_error` as the very first line of any cell containing an intentional error.
"""

from IPython import get_ipython
from IPython.core.magic import register_cell_magic

@register_cell_magic
def allow_error(line, cell):
    # ip.run_cell executes the cell and prints the single native traceback
    get_ipython().run_cell(cell, store_history=False)
```

<h4 style="text-align: center;">Prctice problem on if<h4>

User will enter a floating point number let say 238.915. Your task
is to find out the integer portion before the point (in this case 238)
and then check if that integer portion is an even number or not?

```python
x = float(input("Enter a real number: "))
y = round(x)

if y > x:
    intPortion = y - 1  # e.g., for 29.6, y is 30, intPortion is 30-1=29
else:
    intPortion = y      # e.g., for 29.4, y is 29, intPortion is 29

print(intPortion)
```

```python
x = float(input("Enter a real number: "))
y = round(x)

if x>0:
    if y > x:
        intPortion = y - 1  # e.g., for 29.6, y is 30, intPortion is 30-1=29
    else:
        intPortion = y      # e.g., for 29.4, y is 29, intPortion is 29

else:
    if y < x:
        intPortion = y - 1
    else:
        intPortion = y

if intPortion % 2 == 0:
    print("Even")
else:
    print("Odd")
```

Sort in Assending Order<br>
[1,2,4,-5,7,9,3,2] --> [-5,1,2,2,3,4,7,9]

```python
L = [10,2,4,-5,7,9,3,2,0]
for j in range(len(L)):
    swapped = False
    for i in range(len(L)-j-1):   
        if L[i+1]<L[i]:
            L[i+1],L[i] = L[i], L[i+1]
            swapped = True
        
    if not swapped:
        break
print("Sorted List: ",L)
    
```
